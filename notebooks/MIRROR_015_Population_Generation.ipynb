```python
import numpy as np
import pandas as pd
from pathlib import Path

# Research configuration
SEED = 1926
rng = np.random.default_rng(SEED)

N_HUMANS = 100
N_VARIATIONS = 5

# Locate the project directory
PROJECT_DIR = Path.cwd()

if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

# Establish dataset storage
DATA_DIR = PROJECT_DIR / "datasets"
DATA_DIR.mkdir(exist_ok=True)

# Research initialization
print("MONNIER RESEARCH DIVISION")
print("CAVENDISH GROUP")
print("PROJECT MIRROR — EXPERIMENT 015")
print("-" * 45)

print("Biological sources:", N_HUMANS)
print("Variations per source:", N_VARIATIONS)
print("Total synthetic hosts:", N_HUMANS * N_VARIATIONS)
print("Random seed:", SEED)

print("-" * 45)
print("Dataset location:", DATA_DIR)
print("RESEARCH INITIALIZATION COMPLETE")
```

```python
# Generate randomized biological source population

population_rng = np.random.default_rng(SEED)

# Randomly assign unique biological source IDs
source_ids = population_rng.choice(
    np.arange(1000, 10000),
    size=N_HUMANS,
    replace=False
)

# Establish population names
names = ["Sienna Efendi", "Alexei Sakharov"]

names += [
    f"Anonymous Source {i:03d}"
    for i in range(1, N_HUMANS - 1)
]

# Randomize the population order
population_rng.shuffle(names)

# Generate biological source profiles
humans = []

for i in range(N_HUMANS):

    humans.append({
        "Human_ID": f"HUM-{source_ids[i]}",
        "Human_Name": names[i],
        "Sociability": population_rng.integers(25, 76),
        "Empathy": population_rng.integers(25, 76),
        "Suspicion": population_rng.integers(25, 76),
        "Risk_Tolerance": population_rng.integers(25, 76),
        "Emotional_Attachment": population_rng.integers(25, 76),
        "Adaptability": population_rng.integers(25, 76)
    })

biological_sources = pd.DataFrame(humans)

print("MONNIER — BIOLOGICAL SOURCE REGISTRY")
print("Total registered sources:", len(biological_sources))
print("Unique source IDs:", biological_sources["Human_ID"].nunique())

display(biological_sources.head(10))

# Locate Sienna and Alex within the registry
print("\nSELECTED BIOLOGICAL SOURCES")

display(
    biological_sources[
        biological_sources["Human_Name"].isin(
            ["Sienna Efendi", "Alexei Sakharov"]
        )
    ]
)
```

```python
# Generate synthetic variations

variation_rng = np.random.default_rng(SEED + 1)

TRAITS = [
    "Sociability",
    "Empathy",
    "Suspicion",
    "Risk_Tolerance",
    "Emotional_Attachment",
    "Adaptability"
]

# Initial personality variation
VARIATION_SD = 12

# Established character names
known_hosts = {
    "Sienna Efendi": [
        "Inés Navarro",
        "Nina Valdez",
        "Maya Chen",
        "Clara Voss",
        "Amara Okafor"
    ],
    "Alexei Sakharov": [
        "Elias Haddad",
        "Luca Moretti",
        "Silas Laurent",
        "Rami Nasser",
        "Étienne Moreau"
    ]
}

hosts = []

for _, human in biological_sources.iterrows():

    for v in range(1, N_VARIATIONS + 1):

        mirror_id = f"{human['Human_ID']}-VAR-{v:03d}"

        if human["Human_Name"] in known_hosts:
            host_name = known_hosts[human["Human_Name"]][v - 1]
        else:
            host_name = f"Host-{mirror_id}"

        host = {
            "MIRROR_ID": mirror_id,
            "Human_ID": human["Human_ID"],
            "Host_Name": host_name,
            "Variation": v
        }

        for trait in TRAITS:

            difference = variation_rng.normal(
                loc=0,
                scale=VARIATION_SD
            )

            host[trait] = round(
                np.clip(human[trait] + difference, 0, 100),
                1
            )

        hosts.append(host)

synthetic_hosts = pd.DataFrame(hosts)

print("MONNIER — SYNTHETIC HOST REGISTRY")
print("Total synthetic hosts:", len(synthetic_hosts))
print("Unique host IDs:", synthetic_hosts["MIRROR_ID"].nunique())

display(synthetic_hosts.head(10))

print("\nSELECTED SYNTHETIC VARIATIONS")

display(
    synthetic_hosts[
        synthetic_hosts["Host_Name"].isin(
            known_hosts["Sienna Efendi"] +
            known_hosts["Alexei Sakharov"]
        )
    ]
)
```

```python
# Validate research population

validation = {
    "Correct biological population":
        len(biological_sources) == N_HUMANS,

    "Correct synthetic population":
        len(synthetic_hosts) == N_HUMANS * N_VARIATIONS,

    "Unique biological IDs":
        biological_sources["Human_ID"].is_unique,

    "Unique synthetic IDs":
        synthetic_hosts["MIRROR_ID"].is_unique,

    "All hosts linked to biological sources":
        synthetic_hosts["Human_ID"].isin(
            biological_sources["Human_ID"]
        ).all(),

    "Five variations per biological source":
        synthetic_hosts.groupby("Human_ID").size()
        .reindex(biological_sources["Human_ID"], fill_value=0)
        .eq(N_VARIATIONS).all(),

    "No missing biological data":
        not biological_sources.isnull().any().any(),

    "No missing synthetic data":
        not synthetic_hosts.isnull().any().any(),

    "Valid biological personality scores":
        biological_sources[TRAITS].ge(0).all().all()
        and biological_sources[TRAITS].le(100).all().all(),

    "Valid synthetic personality scores":
        synthetic_hosts[TRAITS].ge(0).all().all()
        and synthetic_hosts[TRAITS].le(100).all().all()
}

print("MONNIER — DATA INTEGRITY AUDIT")
print("-" * 50)

for test, passed in validation.items():
    status = "PASS" if passed else "FAIL"
    print(f"{status} | {test}")

print("-" * 50)
print("Tests passed:", sum(validation.values()), "/", len(validation))

if all(validation.values()):
    print("POPULATION VALIDATED")
else:
    print("DATA INTEGRITY ERROR")
```

```python
# Compare biological and synthetic populations

population_summary = pd.DataFrame({
    "Biological_Mean": biological_sources[TRAITS].mean(),
    "Synthetic_Mean": synthetic_hosts[TRAITS].mean(),
    "Biological_SD": biological_sources[TRAITS].std(),
    "Synthetic_SD": synthetic_hosts[TRAITS].std()
})

population_summary["Mean_Difference"] = (
    population_summary["Synthetic_Mean"] -
    population_summary["Biological_Mean"]
)

print("MONNIER — POPULATION STATISTICS")
display(population_summary.round(2))
```

```python
# Link synthetic hosts to biological sources

comparison = synthetic_hosts.merge(
    biological_sources,
    on="Human_ID",
    suffixes=("_Synthetic", "_Biological"),
    validate="many_to_one"
)

# Calculate individual personality differences

for trait in TRAITS:

    comparison[f"{trait}_Difference"] = (
        comparison[f"{trait}_Synthetic"] -
        comparison[f"{trait}_Biological"]
    )

difference_columns = [
    f"{trait}_Difference"
    for trait in TRAITS
]

# Calculate average absolute personality divergence

comparison["Personality_Divergence"] = (
    comparison[difference_columns]
    .abs()
    .mean(axis=1)
)

print("MONNIER — PERSONALITY DIVERGENCE ANALYSIS")

print("\nAverage personality divergence:")
print(round(comparison["Personality_Divergence"].mean(), 2))

print("\nSIENNA AND ALEX — SYNTHETIC VARIATIONS")

display(
    comparison[
        comparison["Human_Name"].isin(
            ["Sienna Efendi", "Alexei Sakharov"]
        )
    ][[
        "Human_Name",
        "Host_Name",
        "Personality_Divergence"
    ]].round(2)
)
```

```python
print("MONNIER — SAKHAROV DIVERGENCE ANALYSIS")

display(
    comparison[
        comparison["Human_Name"] == "Alexei Sakharov"
    ][[
        "Human_Name",
        "Host_Name",
        "Personality_Divergence"
    ]].sort_values(
        "Personality_Divergence",
        ascending=False
    ).round(2)
)
```

```python
# Export permanent research datasets

assert all(validation.values()), "Population validation failed"

datasets = {
    "biological_sources.csv": biological_sources,
    "synthetic_hosts.csv": synthetic_hosts,
    "personality_comparison.csv": comparison,
    "population_statistics.csv": population_summary
}

print("MONNIER — RESEARCH DATA EXPORT")
print("-" * 50)

for filename, dataset in datasets.items():

    filepath = DATA_DIR / filename

    dataset.to_csv(
        filepath,
        index=(filename == "population_statistics.csv")
    )

    print(f"SAVED | {filename}")
    print(f"Records: {len(dataset)}")
    print()

print("Dataset directory:", DATA_DIR)
print("RESEARCH DATA EXPORT COMPLETE")
```
